```python
# 3) Evaluate the adapter on the same documents
!python -m docextract.llm.infer --model Qwen/Qwen2.5-0.5B-Instruct --adapter runs/qwen05b-lora --name qwen05b-lora-r16 --testset $SETS --limit 200
```
